## Detecting grayscale images

```python
from datasets import load_dataset, Image
from collections import Counter
from PIL import Image as PILImage
import io

ds = load_dataset("Multimodal-Fatima/CUB_train", split="train")
ds = ds.cast_column("image", Image(decode=False))  # raw bytes, skip full decoding

modes, odd = Counter(), []
for i, ex in enumerate(ds):
    im = PILImage.open(io.BytesIO(ex["image"]["bytes"]))  # lazy: reads header only
    modes[im.mode] += 1
    if im.mode != "RGB":
        odd.append((i, im.mode, ex["file_name"]))

print(modes)
print(odd[:20])
```

## Checking Dataset Distribution

```python
from collections import Counter

label_names = ds.features["label"].names
counts = Counter(ds["label"])  # reads only the label column, not the images

print(f"Total rows: {len(ds)} | Classes: {len(counts)}\n")
print(f"{'idx':>3}  {'class':<32} count")
print("-" * 46)
for idx in sorted(counts):
    print(f"{idx:>3}  {label_names[idx]:<32} {counts[idx]}")

vals = list(counts.values())
print(f"\nmin: {min(vals)} | max: {max(vals)} | mean: {sum(vals)/len(vals):.1f}")
```
